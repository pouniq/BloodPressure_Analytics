## Manwhitney U test

I want to take this test to determine whether or not getting blood pressure from left or right hand would differ, and because I have different group sizes from left and right hand I use Manwhitney U Test to get the job done.

```python
from scipy.stats import mannwhitneyu
import pandas as pd
import polars as pl
```

```python
df_R = pl.read_csv('../Data/Right/RightFull.csv') 
df_L = pl.read_csv('../Data/Left/LeftFinal.csv') 
```

```python
names = ['Diastolic', 'Systolic', 'Pulse']
datasets = [df_R, df_L]

threshold = 0.05

for col in names:
    f_stat, p_value = mannwhitneyu(df_R[col],df_L[col])
    print(f'The manwhiteney U test for the {col}: f_statitic: {f_stat} \n the P_value is {p_value:.2f} ')


    if p_value < threshold:
        print(f'The Distribution of right and left {col} measures is not Identical And H_0 is rejected by {p_value:.2f}')
    else:
        print(f'The Distribution of right and left {col} is Identical')
        
    print('-'*50)
```
